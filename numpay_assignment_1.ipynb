# use 1D, 2D and 3D np-array, and apply indexing and slicing atleast 5 example for each (take example as your wish)

```python
x = np.arange(10)
y = np.arange(12).reshape(3, 4)
z = np.arange(27).reshape(3, 3, 3)

print(f"1D Ex1: {x[4:7]}") 
print(f"1D Ex2: {x[:5]}")
print(f"1D Ex3: {x[::2]}")
print(f"1D Ex4: {x[-3:]}")
print(f"1D Ex5: {x[1:8:2]}")

print(f"\n2D Ex1:\n{y[:1, :]}") 
print(f"2D Ex2:\n{y[1:2, 2:4]}") 
print(f"2D Ex3:\n{y[1:, 1:3]}") 
print(f"2D Ex4:\n{y[:, -1]}")
print(f"2D Ex5:\n{y[::2, ::2]}")


print(f"\n3D Ex1:\n{z[0, 1:2, :]}") 
print(f"3D Ex2:\n{z[1, 1:2, :]}") 
print(f"3D Ex3:\n{z[:2, 1:2, 1:3]}") 
print(f"3D Ex4:\n{z[1, 1:3, 1:3]}") 
print(f"3D Ex5:\n{z[:, :, 0]}")
```

# stacking and spliting

```python
d = np.arange(27).reshape(3,9)
e = np.arange(28,55).reshape(3,9)
```

# use hstack() and vstack()

```python

print("hstack:\n", np.hstack((d, e)))

print("vstack:\n", np.vstack((d, e)))
```

# use hsplit() and vsplit()

```python


print(f"hsplit (3 sections):\n{np.hsplit(d, 3)}\n")
print(f"vsplit (3 sections):\n{np.vsplit(e, 3)}")
```
